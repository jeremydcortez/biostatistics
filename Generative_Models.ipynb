```r
# Install and load testthat
install.packages("testthat")
library(testthat)

# Example function to test
add_numbers <- function(a, b) {
  if (!is.numeric(a) || !is.numeric(b)) stop("Inputs must be numeric")
  a + b
}

# Write tests
test_that("add_numbers works as expected", {
  expect_equal(add_numbers(2, 3), 5)           # Checks exact value
  expect_type(add_numbers(2, 3), "double")     # Checks output type
  expect_error(add_numbers("2", 3), "numeric") # Checks error handling
})
```

In the case of HIV, there is an mutation rate of 5 x 10^-4 per nucleotide per replication cycle. HIV genome size is about 10^4. Therefore total # of mutations will follow a Poisson distribution with rate 5 x 10^-4 x 10^ 4 = 5.

Thus, # of mutations over one replication cycle will be close to 5, with variability of $\sqrt{5}$.

To know how often 4 mutations could occur under this model: generate the probability of seeing x= 3 events, take the rate parameter of the distribution $\lambda$ to be 5:

```r
dpois(x = 3, lambda = 5)
```

Therefore the chance of seeing three events is ~0.14

To generate probabilities of all values 0-12 use sequence operator ":".    

```r
dpois(x = 0:12, lambda = 5)

par(bg = "white")
barplot(dpois(0:12, 5), names.arg = 0:12, col = "blue")
```
