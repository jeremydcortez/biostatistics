## Mutation rate example (HIV)

In the case of HIV, there is an mutation rate of 5 x 10^-4 per nucleotide per replication cycle. HIV genome size is about 10^4. Therefore total # of mutations will follow a Poisson distribution with rate 5 x 10^-4 x 10^ 4 = 5.

Thus, # of mutations over one replication cycle will be close to 5, with variability of $\sqrt{5}$.

To know how often 4 mutations could occur under this model: generate the probability of seeing x= 3 events, take the rate parameter of the distribution $\lambda$ to be 5:

```r
dpois(x = 3, lambda = 5)
```

Therefore the chance of seeing three events is ~0.14

To generate probabilities of all values 0-12 use sequence operator ":".    

```r
dpois(x = 0:12, lambda = 5)

par(bg = "white")
barplot(dpois(0:12, 5), names.arg = 0:12, col = "blue")
```

## -- Using discrete prob models --

Point mutations are binary. yes/no. However, not all of biology is binary. Example: genotype in a diploid organism can be three events (AA, Aa, aa).

Other large examples: bacterial types in a sample (100s or 1000s); codons formed of 3 nucleotides (64 levels)

Let's use the blood genotypes for 19 subjects (R calls the variables "factors"); then tabulate:

```r
genotype = c("AA","AO","BB","AO","OO","AO","AA","BO","BO",
             "AO","BB","AO","BO","AB","OO","AB","BB","AO","AO")
table(genotype)
```

You can access what R calls "levels" with the levels function:

```r
genotypeF = factor(genotype)
levels(genotypeF)

table(genotypeF)

par(bg = "white")
freq = table(genotype) 
barplot(freq)
```
